```python
import fitz  # PyMuPDF

def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    texts = []
    for page in doc:
        text = page.get_text()
        if text.strip():  # skip empty pages
            texts.append(text)
    return texts

# Load your PDFs
pdf1_texts = extract_text_from_pdf("/mnt/data/Raven_Johnson_McGraw-Hill_Biology.pdf")
pdf2_texts = extract_text_from_pdf("/mnt/data/Concept-health-Rai-2016.pdf")

# Combine all pages from both PDFs
all_texts = pdf1_texts + pdf2_texts

# Optional: print number of pages read
print(f"Loaded {len(all_texts)} pages from both PDFs.")
```
